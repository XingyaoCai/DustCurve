```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl

# 绘图风格设置
mpl.rcParams.update({'font.size': 18, 'font.family': 'serif', 'axes.linewidth': 2,
                     'xtick.direction': 'in', 'ytick.direction': 'in', 'figure.figsize': (16, 7)})

# ==========================================
# 提取的数据 (根据你提供的值)
# ==========================================
# Redshift Bin 数据
z_centers = np.array([3.4, 4.15, 4.9, 6.25])
f_z = np.array([0.418, 0.414, 0.434, 0.538])
f_z_err = np.array([0.002, 0.004, 0.008, 0.008])

# Dn Bin (Age) 数据
dn_centers = np.array([0.425, 0.625, 0.775, 1.425])
f_dn = np.array([0.356, 0.459, 0.520, 0.518])
f_dn_err = np.array([0.007, 0.005, 0.002, 0.003])

# ==========================================
# 开始绘图
# ==========================================
fig, (ax1, ax2) = plt.subplots(1, 2, sharey=True)

# --- 左图：Redshift 演化 ---
ax1.errorbar(z_centers, f_z, yerr=f_z_err, fmt='o-', color='navy', markersize=10, lw=3, capsize=5, label='Derived $f$-factor')

# 参考线
ax1.axhline(0.44, color='black', linestyle='--', lw=2, alpha=0.7, label='Calzetti (2000): $f=0.44$')
ax1.axhline(0.474, color='purple', linestyle=':', lw=2, alpha=0.7, label='Shivaei (2020): $f=0.474$')

ax1.set_xlabel('Redshift ($z$)', fontsize=22)
ax1.set_ylabel(r'Stellar-to-Nebular Attenuation Ratio $f$', fontsize=22)
ax1.set_xlim(2.8, 6.8)
ax1.set_ylim(0.3, 0.6)
ax1.grid(True, alpha=0.3)
ax1.legend(fontsize=14, loc='lower right')

# --- 右图：Age (Dn) 演化 ---
ax2.errorbar(dn_centers, f_dn, yerr=f_dn_err, fmt='o-', color='firebrick', markersize=10, lw=3, capsize=5)

# 参考线
ax2.axhline(0.44, color='black', linestyle='--', lw=2, alpha=0.7)
ax2.axhline(0.474, color='purple', linestyle=':', lw=2, alpha=0.7)

ax2.set_xlabel(r'Flux Density Ratio $f_{4600}/f_{3400}$', fontsize=22)
ax2.set_xlim(0.2, 1.7)
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('f_factor_Evolution.pdf', dpi=300)
plt.show()
```
