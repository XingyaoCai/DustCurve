```python
# =========================================================================
    # 3. 拟合与分析：多项式拟合 (MC 误差传播)
    # =========================================================================

def run_poly_fitting_mc(self, poly_order=3, n_mc=1000, x_check_range=[2.5, 8.5]):

    wave = np.array(self.effective_curve_wave)
    x_data = 10000.0 / wave
    y_data = self.effective_curve
    y_err = self.effective_curve_err

    mask = np.isfinite(x_data) & np.isfinite(y_data) & np.isfinite(y_err) & (y_err > 0)
    x_fit, y_fit, err_fit = x_data[mask], y_data[mask], y_err[mask]

    # --- A. 解析协方差计算 (Analytical Covariance) ---
    # 使用权重 w=1/sigma 进行加权最小二乘拟合
    weights = 1.0 / err_fit
    coeffs_ana, V = np.polyfit(x_fit, y_fit, poly_order, w=weights, cov=True)

    # 准备计算网格
    x_grid = 10000.0 / np.linspace(1500, 8000, 500)
    # 构建雅可比矩阵 J: 每一行是 [x^n, x^n-1, ..., 1]
    J = np.array([x_grid**i for i in range(poly_order, -1, -1)]).T
    # 计算解析标准差: sigma = sqrt( diag(J * V * J.T) )
    sigma_ana = np.sqrt(np.sum(J * (J @ V), axis=1))

    # --- B. 蒙特卡洛重采样 (与之前一致) ---
    valid_coeffs = []
    for _ in range(n_mc * 5): # 增加采样基数
        y_perturbed = y_fit + np.random.normal(0, err_fit)
        c = np.polyfit(x_fit, y_perturbed, poly_order)
        # 单调性约束
        x_t = np.linspace(x_check_range[0], x_check_range[1], 20)
        if np.all(3*c[0]*x_t**2 + 2*c[1]*x_t + c[2] >= -0.01):
            valid_coeffs.append(c)
            if len(valid_coeffs) >= n_mc: break


    self.mc_coeffs = np.array(valid_coeffs)

    # 1. 建立高精度的波长网格用于生成曲线
    x_grid = 10000.0 / np.linspace(1500, 8000, 500)

    # 2. 生成所有合法的曲线族 (形状: n_mc x 500)
    all_curves = np.array([np.polyval(c, x_grid) for c in self.mc_coeffs])

    # 3. 【核心修改】在每个波长点上独立计算百分位数
    # 这样能保证 Best Fit 严格遵循筛选后的样本分布
    best_fit_curve = np.percentile(all_curves, 50.0, axis=0) # 中位数曲线
    low_1sig = np.percentile(all_curves, 16.0, axis=0)
    high_1sig = np.percentile(all_curves, 84.0, axis=0)

    # 4. 同时更新类的属性，确保系数也是从中位数样本中提取
    # 注意：多项式系数的中位数生成的曲线，与 曲线的中位数 略有不同
    # 我们以曲线的中位数为准绘制蓝线
    self.poly_coeffs = np.median(self.mc_coeffs, axis=0)

    self.fit_results = {
            'wave_grid': 10000.0 / x_grid,
            'best_fit': best_fit_curve,  # 这里的蓝线现在是 50th 百分位
            'low_1sig_mc': low_1sig,
            'high_1sig_mc': high_1sig,
            'sigma_ana': sigma_ana,      # 协方差误差仅作为数学参考
            'sigma_data': np.std(y_fit - np.polyval(self.poly_coeffs, x_fit))
        }
def plot_attenuation_results(self, xlim=(1500, 8000), ylim=(-1, 3)):
    res = self.fit_results
    plt.figure(figsize=(16, 12))
    plt.rcParams['text.usetex'] = False

    # 1. 原始数据
    plt.errorbar(self.effective_curve_wave[::10], self.effective_curve[::10],
                    yerr=self.effective_curve_err[::10],
                    fmt='o', color='black', alpha=0.2, markersize=3, label='Data', capsize=0)

    # 2. MC 置信区间 (反映了单调性约束后的真实分布)
    plt.fill_between(res['wave_grid'], res['low_1sig_mc'], res['high_1sig_mc'],
                        color='royalblue', alpha=0.3, label='MC 1$\sigma$ (with Constraints)')

    # 3. 协方差解析误差 (Analytical 1-sigma)
    # 我们用虚线表示，看看它和 MC 结果是否重合
    plt.plot(res['wave_grid'], res['best_fit'] + res['sigma_ana'], ':', color='red', lw=1.5,
                label='Analytical 1$\sigma$ (from Covariance)')
    plt.plot(res['wave_grid'], res['best_fit'] - res['sigma_ana'], ':', color='red', lw=1.5)

    # 4. 数据标准差带
    plt.plot(res['wave_grid'], res['best_fit'] + res['sigma_data'], '--', color='gray', alpha=0.4, lw=1,
                label=f'Data Scatter (SD={res["sigma_data"]:.2f})')
    plt.plot(res['wave_grid'], res['best_fit'] - res['sigma_data'], '--', color='gray', alpha=0.4, lw=1)

    # 5. 最佳拟合线
    plt.plot(res['wave_grid'], res['best_fit'], color='navy', lw=2.5, label='Best Poly Fit')

    plt.xlabel('Wavelength ($\AA$)', fontsize=13)
    plt.ylabel(r'$Q_\lambda$', fontsize=13)
    plt.legend(loc='best', frameon=True, fontsize=9)
    plt.ylim(np.percentile(res['low_1sig_mc'], 5)-0.5, np.percentile(res['high_1sig_mc'], 95)+0.5)
    plt.show()
```

```python
def calculate_statistics(self):
    """
    计算拟合的 Chi-squared (kai2) 和 Reduced Chi-squared。
    """
    if not hasattr(self, 'poly_coeffs'):
        print("Please run run_mcmc_fitting first.")
        return

    # 1. 准备数据 (确保与拟合时一致)
    wave_um = self.effective_curve_wave / 10000.0
    x_data = 1.0 / wave_um
    y_data = self.effective_curve
    y_err = self.effective_curve_err

    # 剔除无效数据
    mask = np.isfinite(x_data) & np.isfinite(y_data) & np.isfinite(y_err) & (y_err > 0)
    x_fit, y_obs, err_obs = x_data[mask], y_data[mask], y_err[mask]

    # 2. 计算模型预测值
    # 使用 MCMC 得到的最佳系数 (通常是中位数)
    y_model = np.polyval(self.poly_coeffs, x_fit)

    # 3. 计算 Chi-squared
    # 公式: sum( (obs - model)^2 / err^2 )
    chi2 = np.sum(((y_obs - y_model) / err_obs)**2)

    # 4. 计算自由度 (Degrees of Freedom, dof)
    # 公式: 数据点数量 - 拟合参数数量
    n_data = len(y_obs)
    n_params = len(self.poly_coeffs)
    dof = n_data - n_params

    # 5. 计算简缩 Chi-squared (Reduced Chi-squared)
    chi2_red = chi2 / dof

    print(f"\n--- Fit Statistics (kai2) ---")
    print(f"Number of data points (N): {n_data}")
    print(f"Number of parameters (k): {n_params}")
    print(f"Degrees of freedom (dof): {dof}")
    print(f"Total Chi-squared (kai2): {chi2:.4f}")
    print(f"Reduced Chi-squared (kai2_red): {chi2_red:.4f}")

    # 解释结果
    if chi2_red > 2:
        print("结论: kai2_red >> 1。模型未能完全解释观测波动（可能欠拟合，或误差被低估）。")
    elif 0.5 <= chi2_red <= 2:
        print("结论: kai2_red ≈ 1。拟合非常理想，误差分量与观测噪声匹配。")
    else:
        print("结论: kai2_red << 1。拟合过于‘完美’（可能过拟合，或观测误差被严重高估）。")

    return chi2, dof, chi2_red
```

```python
import numpy as np

def calculate_dual_Rv_mcmc(analyst):
    """
    根据 MCMC 链，分别计算 k(infinity)=0 和 k(2.85um)=0 两种假设下的 R_V 及其误差。
    """
    if not hasattr(analyst, 'mcmc_samples'):
        raise ValueError("Please run analyst.run_mcmc_fitting() first.")

    samples = analyst.mcmc_samples

    # 关键波长 (单位: um^-1)
    x_V = 1.0 / 0.55
    x_B = 1.0 / 0.44
    x_285 = 1.0 / 2.85
    x_inf = 0.0

    Rv_inf_list = []
    Rv_285_list =[]

    for coeffs in samples:
        # coeffs = [c3, c2, c1, c0]

        # 1. 计算各个波长处的形状多项式 P(x) 的值
        P_V = np.polyval(coeffs, x_V)
        P_B = np.polyval(coeffs, x_B)
        P_285 = np.polyval(coeffs, x_285)
        P_inf = np.polyval(coeffs, x_inf) # 实际上就是 c0

        # 2. 计算 Scaling factor S
        # 强制 k(B) - k(V) = 1  => S * (P_B - P_V) = 1
        S = 1.0 / (P_B - P_V)

        # 3. 假设 1: k(infinity) = 0
        # k(x) = S * P(x) + C_inf
        # k(inf) = S * P_inf + C_inf = 0 => C_inf = -S * P_inf
        # R_V = k(V) = S * P_V + C_inf
        C_inf = -S * P_inf
        Rv_inf = S * P_V + C_inf
        Rv_inf_list.append(Rv_inf)

        # 4. 假设 2: k(2.85um) = 0 (Reddy et al. 2015)
        # k(x) = S * P(x) + C_285
        # k(2.85) = S * P_285 + C_285 = 0 => C_285 = -S * P_285
        # R_V = k(V) = S * P_V + C_285
        C_285 = -S * P_285
        Rv_285 = S * P_V + C_285
        Rv_285_list.append(Rv_285)

    Rv_inf_list = np.array(Rv_inf_list)
    Rv_285_list = np.array(Rv_285_list)

    # 统计计算 (中位数与 1-sigma 误差)
    res_inf = {
        'med': np.median(Rv_inf_list),
        'up': np.percentile(Rv_inf_list, 84) - np.median(Rv_inf_list),
        'low': np.median(Rv_inf_list) - np.percentile(Rv_inf_list, 16)
    }

    res_285 = {
        'med': np.median(Rv_285_list),
        'up': np.percentile(Rv_285_list, 84) - np.median(Rv_285_list),
        'low': np.median(Rv_285_list) - np.percentile(Rv_285_list, 16)
    }

    print("-" * 50)
    print(f"Assumption 1: k(inf) = 0")
    print(f"R_V = {res_inf['med']:.3f} + {res_inf['up']:.3f} / - {res_inf['low']:.3f}")
    print("-" * 50)
    print(f"Assumption 2: k(2.85um) = 0 (Reddy et al. 2015)")
    print(f"R_V = {res_285['med']:.3f} + {res_285['up']:.3f} / - {res_285['low']:.3f}")
    print("-" * 50)
    print(f"Systematic Difference (Delta R_V): {res_inf['med'] - res_285['med']:.3f}")

    return res_inf, res_285

# 运行方法：将你每个 Bin 的 analyst 传入即可，例如：
# res_inf, res_285 = calculate_dual_Rv_mcmc(analyst)


for i, analyst in enumerate(analyst_list):
    R_v_list=[]
    wavelengths = np.arange(1250, 30000, 10)
    calzetti=2.659*analyst.get_calzetti_law(wavelengths)+4.05
    own_law=analyst._poly_shape(10000/wavelengths)
    normalization_own=1/(analyst._poly_shape(10000/4400)-analyst._poly_shape(10000/5500))
    R_v=-normalization_own*analyst._poly_shape(10000/28500)
    own_law_normalized=own_law*normalization_own+R_v

    # plt.figure(figsize=(10, 6))
    # plt.plot(wavelengths, calzetti, label='Calzetti Law', color='red')
    # plt.plot(wavelengths, own_law_normalized, label='Own Law', color='blue')
    for sample in analyst.mcmc_samples[np.random.choice(analyst.mcmc_samples.shape[0], 100, replace=False)]:
        c3, c2, c1, c0 = sample
        sample_law = c3*(10000/wavelengths)**3 + c2*(10000/wavelengths)**2 + c1*(10000/wavelengths) + c0
        normalization_samle=1/(c3*(10000/4400)**3 + c2*(10000/4400)**2 + c1*(10000/4400) + c0 - (c3*(10000/5500)**3 + c2*(10000/5500)**2 + c1*(10000/5500) + c0))
        R_v_sample=-normalization_samle*(c3*(10000/28500)**3 + c2*(10000/28500)**2 + c1*(10000/28500) + c0)
        R_v_list.append(R_v_sample)
        sample_law_normalized = sample_law*normalization_samle + R_v_sample
        # plt.plot(wavelengths, sample_law_normalized, color='blue', alpha=0.1)
    # plt.xlabel('Wavelength (Å)')
    # plt.ylabel('Attenuation (k(λ))')
    # plt.title('Comparison of Attenuation Curves')
    # plt.legend()
    # plt.grid()
    # plt.show()
    print(f"Derived Rv (from samples): {np.mean(R_v_list):.4f} ± {np.std(R_v_list):.4f}")
    print(f"Derived Rv (from median and 1-sigma CI): {np.median(R_v_list):.4f} ± {np.percentile(R_v_list, 99)-np.median(R_v_list):.4f} / {np.median(R_v_list)-np.percentile(R_v_list, 1):.4f}")
    print('='*50)
    redshift_range = redshift_conditions[i]
    condition_name = f"Z {redshift_range[0]:.2f}-{redshift_range[1]:.2f}"
    print(f"\nCondition: {condition_name}")
    f_mean, f_median, f_err_upper, f_err_lower = analyst.derive_f_factor()
    print(f"Derived f factor: {f_mean:.4f} (Median: {f_median:.4f}, 1-sigma CI: [{f_err_lower:.4f}, {f_err_upper:.4f}])")
    print('='*50)
    res_inf, res_285 = calculate_dual_Rv_mcmc(analyst)
    print('='*50)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patheffects as pe

def custom_plot_reddening_curve_comparison(analyst,
                                           if_show=True,
                                           if_input=False,
                                           fig=None,
                                           ax=None,
                                           curve_label='This Study',
                                           curve_color='#002147',
                                           draw_background=True,
                                           plot_mcmc_samples=False,
                                           n_samples=100):
    """
    外部重写的绘制红化对比图函数。
    【修改内容】：MCMC 样本线使用主线同色，且不再进入图例。
    """
    if analyst.scaling_factor is None:
        analyst.derive_k_curve_parameters()

    if if_input:
        if fig is None or ax is None:
            raise ValueError("If if_input is True, fig and ax must be provided.")
    else:
        fig, ax = plt.subplots(figsize=(12, 10))
        draw_background = True

    x_grid = np.linspace(0.5, 8.2, 1000) # 1/um

    # ==========================================
    # 3. 准备 MCMC 样本线
    # ==========================================
    sample_lines_y =[]
    if plot_mcmc_samples and hasattr(analyst, 'mcmc_samples'):
        x_B, x_V = 1.0/0.44, 1.0/0.55
        n_total = len(analyst.mcmc_samples)

        actual_n = min(n_samples, n_total)
        inds = np.random.choice(n_total, size=actual_n, replace=False)

        for ind in inds:
            coeffs = analyst.mcmc_samples[ind]
            c3, c2, c1 = coeffs[0], coeffs[1], coeffs[2]

            def shape_func(x):
                return c3 * x**3 + c2 * x**2 + c1 * x

            val_grid = shape_func(x_grid)
            val_B = shape_func(x_B)
            val_V = shape_func(x_V)

            y_sample = (val_grid - val_V) / (val_B - val_V)
            sample_lines_y.append(y_sample)

    # ==========================================
    # 4. 计算最佳拟合曲线 (Best Fit)
    # ==========================================
    shape_grid = analyst._poly_shape_cubic_only(x_grid)
    shape_V = analyst._poly_shape_cubic_only(1.0/0.55)
    shape_B = analyst._poly_shape_cubic_only(1.0/0.44)
    y_derived = (shape_grid - shape_V) / (shape_B - shape_V)

    # ==========================================
    # 5. 绘制背景参考线
    # ==========================================
    if draw_background:
        y_calz = analyst._get_normalized_reddening(x_grid, 'calzetti')
        y_mw = analyst._get_normalized_reddening(x_grid, 'mw')
        y_smc = analyst._get_normalized_reddening(x_grid, 'smc')

        ax.plot(x_grid, y_mw, color='#1f77b4', ls=':', lw=3, label='Milky Way (CCM89)')
        ax.plot(x_grid, y_smc, color='#2ca02c', ls='-.', lw=3, label='SMC (Pei92)')
        ax.plot(x_grid, y_calz, color='#d62728', ls='--', lw=3, label='Calzetti (2000)')

        # 辅助线和标记
        x_V, x_B = 1.0/0.55, 1.0/0.44
        ax.axhline(0, color='gray', linestyle='-', alpha=0.3)
        ax.axvline(x_V, color='gray', linestyle='-', alpha=0.3)
        ax.scatter([x_V], [0], color='black', s=80, zorder=10)
        ax.text(x_V, 0.5, r'V ($5500\,\mathrm{\AA}$)', rotation=90, color='gray', fontsize=20, ha='right')

        ax.axhline(1, color='gray', linestyle=':', alpha=0.3)
        ax.axvline(x_B, color='gray', linestyle=':', alpha=0.3)
        ax.scatter([x_B], [1], color='black', s=80, zorder=10)
        ax.text(x_B, 1.5, r'B ($4400\,\mathrm{\AA}$)', rotation=90, color='gray', fontsize=20, ha='right')

        # 顶部波长轴
        ax2 = ax.twiny()
        ticks =[10000, 5500, 4400, 3000, 2000, 1500]
        ax2.set_xlim(0.5, 8.2)
        ax2.set_xticks([10000.0/t for t in ticks])
        ax2.set_xticklabels([str(t) for t in ticks], fontsize=18)
        ax2.set_xlabel(r'Wavelength $\lambda\ (\mathrm{\AA})$', fontsize=28)

        # 开启次级刻度
        ax.minorticks_on()
        ax.tick_params(axis='both', which='major', length=8, width=1.5, direction='in', right=True)
        ax.tick_params(axis='both', which='minor', length=4, width=1, direction='in', right=True)
        ax2.minorticks_on()
        ax2.tick_params(axis='x', which='major', length=8, width=1.5, direction='in')
        ax2.tick_params(axis='x', which='minor', length=4, width=1, direction='in')

        ax.grid(True, alpha=0.2)

    # ==========================================
    # 6. 绘制 MCMC 样本线 (在底层)
    # ==========================================
    if sample_lines_y:
        for y_samp in sample_lines_y:
            ax.plot(x_grid, y_samp,
                    color=curve_color,   # 【修改点】：直接使用传入的主线颜色
                    alpha=0.03,          # 透明度保持极低
                    lw=1.5,
                    zorder=1)            # 【修改点】：不再包含 label 属性

    # ==========================================
    # 7. 绘制最佳拟合线 (带有白色外轮廓 alpha=0.25)
    # ==========================================
    ax.plot(x_grid, y_derived,
            color=curve_color,
            lw=4,
            label=curve_label,
            zorder=5,
            path_effects=[pe.Stroke(linewidth=6, foreground='white', alpha=0.5), pe.Normal()])

    if if_show:
        # 单独显示时的逻辑
        ax.set_xlim(0.5, 8.2)
        ax.set_xlabel(r'Inverse Wavelength $\lambda^{-1}\ (\mu\mathrm{m}^{-1})$', fontsize=28)
        ax.set_ylabel(r'$E(\lambda - V) / E(B - V)$', fontsize=28)

        handles, labels = ax.get_legend_handles_labels()
        by_label = dict(zip(labels, handles))
        ax.legend(by_label.values(), by_label.keys(), loc='upper left', framealpha=0.9, edgecolor='none')
        plt.tight_layout()
        plt.show()

    return fig, ax
```
