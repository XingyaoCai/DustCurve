```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib as mpl

# 绘图风格设置
mpl.rcParams.update({'font.size': 18, 'font.family': 'serif', 'axes.linewidth': 2,
                     'xtick.direction': 'in', 'ytick.direction': 'in', 'figure.figsize': (16, 7)})

# ==========================================
# 提取的数据
# ==========================================
# Redshift Bin 数据
z_centers = np.array([3.4, 4.15, 4.9, 6.25])
z_rv_inf = np.array([4.960, 5.862, 8.404, 7.999])
z_rv_inf_err = np.array([[0.026, 0.096, 0.111, 0.062],[0.027, 0.092, 0.107, 0.053]]) # [lower, upper]
z_rv_285 = np.array([3.878, 4.468, 6.123, 5.866])
z_rv_285_err = np.array([[0.017, 0.062, 0.072, 0.040],[0.018, 0.059, 0.070, 0.034]])

# Dn Bin (Age) 数据
dn_centers = np.array([0.425, 0.625, 0.775, 1.425]) # 最后一个bin比较宽，画在中间偏右
dn_rv_inf = np.array([7.847, 7.236, 5.087, 5.106])
dn_rv_inf_err = np.array([[0.106, 0.082, 0.017, 0.031],[0.103, 0.083, 0.018, 0.032]])
dn_rv_285 = np.array([5.764, 5.364, 3.962, 3.974])
dn_rv_285_err = np.array([[0.069, 0.053, 0.012, 0.020],[0.066, 0.054, 0.012, 0.021]])

# ==========================================
# 开始绘图
# ==========================================
fig, (ax1, ax2) = plt.subplots(1, 2, sharey=True)

# --- 左图：Redshift 演化 ---
ax1.errorbar(z_centers, z_rv_inf, yerr=z_rv_inf_err, fmt='o-', color='navy', markersize=10, lw=3, capsize=5, label=r'Assume $k(\infty)=0$')
ax1.errorbar(z_centers, z_rv_285, yerr=z_rv_285_err, fmt='s--', color='firebrick', markersize=10, lw=3, capsize=5, label=r'Assume $k(2.85\mu m)=0$ (Reddy+15)')
ax1.axhline(4.05, color='black', linestyle=':', lw=2, alpha=0.6, label='Calzetti (2000) $R_V=4.05$')

ax1.set_xlabel('Redshift ($z$)', fontsize=22)
ax1.set_ylabel('Total-to-selective Extinction $R_V$', fontsize=22)
ax1.set_xlim(3.0, 6.8)
ax1.set_ylim(2, 9)
ax1.grid(True, alpha=0.3)
ax1.legend(fontsize=14)

# --- 右图：Age (Dn) 演化 ---
ax2.errorbar(dn_centers, dn_rv_inf, yerr=dn_rv_inf_err, fmt='o-', color='navy', markersize=10, lw=3, capsize=5)
ax2.errorbar(dn_centers, dn_rv_285, yerr=dn_rv_285_err, fmt='s--', color='firebrick', markersize=10, lw=3, capsize=5)
ax2.axhline(4.05, color='black', linestyle=':', lw=2, alpha=0.6)

ax2.set_xlabel(r'Flux Density Ratio $f_{4600}/f_{3400}$', fontsize=22)
ax2.set_xlim(0.3, 1.6)
# ax2.set_xscale('log')
ax2.grid(True, alpha=0.3)

plt.tight_layout()
plt.savefig('Rv_Evolution.pdf', dpi=300)
plt.show()
```
